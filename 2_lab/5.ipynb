## Оператор Try-except
- це є конструкція опрацьовувача помилок, яка дозволяє обробляти винятки, що виникають під час виконання програми. Вона складається з блоку `try`, де розміщується код, який може викликати помилку, та блоку `except`, де визначається, як обробляти цю помилку.
- яскравим прикладо є те що ділити на 0 неможна, при діленні виникає помилка `ZeroDivisionError`, яку можна обробити за допомогою конструкції try-except. 

```python
import random

list_of_numbers = list(range(-10, 10)) # 20 елементів і серед них 0
# Нуль може випасти або і не випасти, тому помилка ж випадковою
for _ in range(10):
    na = random.choice(list_of_numbers)
    # для того щоб виловити і правильно обробити помилку ділення на 0, 
    # використаємо конструкцію try-except
    try:
        # Це блок коду, який може викликати помилку
        # якщо помилки немає, блок коду виконується нормально, і програма продовжує виконання
        print(f"Ділення на {na} дає результат: ", 1/na)
    except ZeroDivisionError:
        # Це блок коду, який виконується, якщо виникає помилка ZeroDivisionError
        print("Обролювач помилки: Ділення на нуль неможливе!")
        print(f"Ділення на {na} неможливе, результат: ", None)
```

- код внизу буде давати різні пимилки, але якщо виконати try-except, програма не зупиниться і буде виконуватися далі.

```python
try:
    None + 1
    #import boto3
    #print(list_of_numbers[20])
except Exception as e:
    print(f"Помилка: {e}")
```

## Контекст менеджер with
- він відповідає за узгоджене відкриття та закриття ресурсів, таких як файли, мережеві з'єднання або бази даних. Він забезпечує автоматичне закриття ресурсу після завершення роботи з ним, навіть якщо виникає помилка під час виконання коду всередині блоку `with`. Це допомагає уникнути витоків ресурсів та забезпечує більш чистий і безпечний код.

```python
fa = open("file.txt", "w")
fa.write("~Перший запис~")

fb = open("file.txt", "a") # fb - це вказівник на файл, який відкритий для додавання (append)
fb.write("+Записали Другий файл+")

fb.close()
fa.close()
```

```python
with open("file_2.txt", "w") as f:
    f.write("~Перший запис~")
    f.write("+Записали Другий файл+")
    # після виходу з контекстного менеджера файл автоматично закривається
# при повторному відкриттін файлу, конфліків небуде, 
# бо файл менеджиться контекстним менеджером
with open("file_2.txt", "a") as f:
    f.write("\n\n\n")
    f.write("Цей рядок запишеться бо другий раз відрили файл")
```
