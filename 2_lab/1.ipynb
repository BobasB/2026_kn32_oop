### Починаємо з типів даних
Python має кілька вбудованих типів даних, включаючи:
- **int**: Цілі числа (наприклад, 1, 42)
- **float**: Числа з плаваючою комою (наприклад, 3.14, 2.0)

```python
a = 1.0
b = 2
c = a + b
print(c)
```

- **str**: стрічкові типи даних

```python
g = '1'
h = "2"
s = "Це є стрічка"
print(s)
```

- casting - явне перетворення типів даних (наприклад, `int()` для перетворення в ціле число, `str()` для перетворення в рядок)
- є явне та неявне перетворення типів даних.

```python
casting_str = int(g)
print("Ми маємо змінну з величиною ", casting_str, " яка є типу ", type(casting_str), " ти змінну ", g, " яка є типу ", type(g))
```

- арифметичні операції та перетворення типів даних

```python
d = a + b
print("Це проста арифметична операція: ", d)
# дана операція буде помилковою, оскільки ми намагаємося додати рядок та число
#j = d + h

# для того, щоб додати рядок та число, потрібно перетворити стрічку в число
j = d + int(h)
print("Тепер все працює правильно: ", j)
```

- перевантаження оператора - це механізм, який дозволяє визначити поведінку оператора для користувача визначених типів даних.
- на стрічкових даних буде називатися конкатенацією, а на числових - додаванням.

```python
s1 = "Ми вчимось "
s2 = "мові Python"
s3 = s1 + s2
print(s3)
```

- колекції (набори) - це структури даних, які дозволяють зберігати кілька значень в одному об'єкті.
- колекції мають впорядковані дані які стоять за певних індексом
- індекси починаються з нуля
- індекси можуть бути від'ємними, що дозволяє звертатися до елементів з кінця колекції
