### Вбудовані функції

```python
a = 1
b = str(a)
print(b, type(b))
```

детально пропрацюємо роботу вбудованої функції print() - вона виводить на екран значення, які передані їй у вигляді аргументів.
- в найпростішому вигляді - просто передаємо аргуменнти для виводу
- аргумент - це значення відделене комою, яке передається у функцію для обробки

```python
print(1, 2, 3, 4) # ми передали 4 аргументи у функцію print() - вона вивела їх на екран, розділивши пробілом
print(1, 1.1, " 1 ", "Один") # ми передали 4 аргументи різних типів даних
# між аргументами функція print() автоматично вставляє пробіл
print("Це довга стрічка з пробілами", "і тут різділено пробулом", "і тут теж пробіл")
s1 = "Це довга стрічка з пробілами"
s2 = "і продовження цієї стрічки"
print(s1, s2) # ми передали 2 аргументи у функцію
# при конкатенації рядків (стрічок) функція print() не вставляє пробіл між ними, якщо ми його не додамо самі
s3 = s1 + s2
print(s3) # ми передали 1 аргумент у функцію
s4 = s1 + " " + s2 # конкатенація рядків з явним пробілом між ними
print(s4) # ми передали 1 аргумент у функцію
```

- в час АІ гнучкість з роботою з стічками дає багато переваг, наприклад, можна передавати у функцію print() не тільки стічки, а й інші типи даних, які будуть автоматично перетворені у стічку для виводу на екран. 

```python
p1 = "# Системний промпт"
p2 = "Ти є спеціалістом з Python, який допомагає студентам вивчати мову програмування Python"
p3 = "# Правила для написання тестовийх завдань"
p4 = """- проста складність завдань
- короткі завдання
- завдання на основі прикладів
"""
p5 = "# Приклади завдань"

final_prompt = p1 + "\n" + p2 + "\n\n\n" + p3 + "\n" + p4 + "\n---\n" + p5 + "\n"
print(final_prompt)
```

З того, що ми пройшли, вже видно що можемо формувати стрічку до її виводу або закидати аргументи у функцію print() і вона сама перетворить їх у стрічку для виводу на екран.

у Python для формування стрічок використовується властивості str - це методи, які дозволяють виконувати різні операції над стрічками. Методи викликаються через крапку після стрічки, наприклад: `str.upper()`, `str.lower()`, `str.replace()`, `str.split()`, `str.join()` а також `.format()` - це метод, який дозволяє формувати стрічку з різних значень, які передані у вигляді аргументів.  

```python
so = "Це довга стрічка {} і продовження цієї стрічки {} і в кінці {}".format("ТУТ", "ТАМКАЙ", "ТУТ")
print(so)
sn = "Це довга стрічка {0} і продовження цієї стрічки {1} і в кінці {0}".format("ТУТ", "ТАМКАЙ")
print(sn)
```

зараз найбільш поширеним методом формування стрічок є f-стрічки, які дозволяють формувати стрічку з різних значень, які передані у вигляді аргументів.

```python
fs = f"Ця стрічка пуста і не форматована" # ця F-string не містить змінних, тому вона не форматована
print(fs)
af = "ВСТАВКА!"
fs_1 = f"Ця стрічка містить змінну fs = {fs} та додакові дані = {af}"

fs_2 = f"А тепер все разом: {fs_1}"

print(fs_2)
```

реальний приклад використання f-стрічок - простий email, який ми можемо сформувати з різних значень, які передані у вигляді аргументів.

```python
name = "Богдан"
greeting = f"Привіт, {name}!"
main_text = f"Запрошуємо на вивчення ООП з Python. {name}, не пропусти шанс стати експертом у Python!"
footer = f"З повагою. ІТ коледж"
filler = '~|~'*10

# шаблон, темплейт, скретч - вживаємо будь-який з цих термінів, 
# щоб позначити заготовку, яку ми можемо заповнити змінними
email_template = f"""
{filler}
{greeting}

{main_text}

---
{footer}

{filler}
"""

print(email_template)
```

Цікавою вбудованою функцією є map - вона дозволяє застосувати функцію до кожного елементу ітерабельного об'єкта (наприклад, списку) і повертає новий ітерабельний об'єкт з результатами.

```python
mp = map(print, ["Це перший елемент", "Це другий елемент", "Це третій елемент"])

list(mp) # ми перетворюємо об'єкт map на список, щоб викликати функцію print() для кожного елементу списку
```
