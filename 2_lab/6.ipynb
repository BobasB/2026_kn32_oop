## Функції та лямбда вирази
- головне правило програмування: "не повторюй себе" (DRY - Don't Repeat Yourself) - будь-який код який дублюється, слід винести в окрему функцію або метод.
- якщо по простому, то функція це блок коду який ма є назву, може приймати аргументи і повертати результат. І по назві функції ми можемо викликати цей код. Функції дозволяють структурувати код, роблять його більш читабельним та зручним для повторного використання.
- функці при виклику має мати дужки () після назви, навіть якщо функція не приймає аргументів.

```python
def my_function_name_no_args():
    print("Функція виводить текст без аргументів")
    print("Ще один рядок тексту")
    print("Завершили блок коду")

my_function_name_no_args()
print(type(my_function_name_no_args))
```

- оскільки не викликана функція не виконує код то ми можемо її присвоювати змінній або передавати в іншу функцію як аргумент. 

```python
def my_function_with_return_value():
    return "Функція повертає рядок тексту"

a = my_function_with_return_value()
b = my_function_with_return_value

print(a, b, b())
```

- можна зробити хитру штуку, передати всередину у функцію іншу функцію і викликати її всередині.  

```python
def my_fun_1():
    return "Функція 1"

def my_fun_2(f):
    return f

def my_fun_with_function_arguments(function_argument):
    return function_argument()

print(my_fun_with_function_arguments(my_fun_2(my_fun_1)))
```

- іменнні та безіменні функції (лямбда вирази) - це функції які можна створювати без імені, вони можуть приймати аргументи і повертати результат. Лямбда вирази використовуються для створення невеликих функцій, які використовуються один раз або передаються як аргументи в інші функції.

```python
def my_classic_function(x):
    """Іменнна функція яка містить вираз"""
    return (x * 2) ** 2 if x > 0 else 0

a = [my_classic_function(x) for x in range(-5, 5)]
print(a)

# замінюємо на лямбду
b = [(lambda x: (x * 2) ** 2 if x > 0 else 0)(x) for x in range(-5, 5)]
print(b)
```
