## Розгалуження
- класично це є if else конструкція, яка дозволяє виконувати різні блоки коду залежно від умови.
- в ІТ майже не користуються конструкцією if else, а використовують скоротні форми

```python
import random
condition = random.choice([True, False])

if condition:
    print("Умова істинна, виконується цей блок коду.")
else:
    print("Умова хибна, виконується інший блок коду.")
```

- якщо нам потрібно виконати операцію тільки при певній умові, то ми можемо використати скорочену форму if:

```python
c = False
print("1. Цей код виконується завжди, незалежно від умови.")

if c:
    print("2. Ця умова лише при істині.")

print("3. Цей код виконується завжди, незалежно від умови.")
```

- для складних багаторівневих умов можна використовувати конструкцію if elif else, яка дозволяє перевіряти кілька умов підряд.

```python
rainbow = ["червоний", "оранжевий", "жовтий", "зелений", "блакитний", "синій", "фіолетовий"]
color = random.choice(rainbow)

# це класичний запис але так не роблять в ІТ, бо це громіздко і не зручно
# якщо випадає останній колір ми маємо всеодно пройти перевірку всіх попередніх умов
if color == "червоний":
    print("Перший колір веселки.")
elif color == "оранжевий":
    print("Другий колір веселки.")
elif color == "жовтий":
    print("Третій колір веселки.")
elif color == "зелений":
    print("Четвертий колір веселки.")
elif color == "блакитний":
    print("П'ятий колір веселки.")
elif color == "синій":
    print("Шостий колір веселки.")
else:
    print("Останній колір веселки.")

# це вже краще бо це компактно і зручно
# ми можемо використати словник, де ключами будуть кольори, а значеннями - їх опис
# тут здійснюється пошук ключа в словнику за одну операцію, а не перевірка всіх умов підряд
d_rainbow = {
    "червоний": "Перший колір веселки.",
    "оранжевий": "Другий колір веселки.",
    "жовтий": "Третій колір веселки.",
    "зелений": "Четвертий колір веселки.",
    "блакитний": "П'ятий колір веселки.",
    "синій": "Шостий колір веселки.",
    "фіолетовий": "Останній колір веселки."
}

print(d_rainbow.get(color, "Колір не знайдено у веселці."))
```

- if else конструкція часто поєднується у функція з оператором return, що дозволяє повертати значення залежно від умови.
- за менше кількість коду викнуємо ту ж саму логіку, що і у класичному if else.

```python
# класичний запис з використанням if else конструкції
def function_name(x):
    if x > 0:
        return "Позитивне число"
    else:
        return "Негативне число або нуль"

# скорочений запис з використанням тернарного оператора
def function_name_short(x):
    if x > 0:
        return "Позитивне число"
    # якщо x більше 0 то сюди ми ніколи не дійдемо, бо ми вже повернули значення з функції
    return "Негативне число або нуль"
```
