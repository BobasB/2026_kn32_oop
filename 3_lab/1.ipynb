## Обєктно-орієнтоване програмування (ООП)

- приклад-порівняння чому нам не достатньо. вбудованих класів та функцій, і чому нам потрібні власні класи та функції.
- описуємо студента та його оцінки.

```python
# Використовуємо вбудовані типи, словник, який є обєктом, що зберігає пари ключ-значення. 
student = {
    "Богдан": {"Вища математика": 4, "фізика": 5, "ООП": 5},
    "Андрій": {"Вища математика": 5, "фізика": 4},
    "Анастісія": {"Вища математика": 4, "фізика": 4},
}

print(type(student))  # <class 'dict'>

for name, grade in student.items():
    print(type(grade))  # <class 'dict'>
    print(f"{name} отримав оцінки:")
    print(f"\tВища математика: {grade['Вища математика']}")
    print(f"\tФізика: {grade['фізика']}")
    print(f"\tООП: {grade.get('ООП', 'немає оцінки')}\n") # метод get() дає значення за замовчуванням
```

- використання вбудованих типів для складних задач швидко стає громіздкою та незручною (не маштабованим).
- тому - ООП підхід дає можливість створювати власні класи та функції, які будуть більш зручними та маштабованими для вирішення складних задач.

- клас свторюється за допомогою ключового слова class, а обєкти створюються за допомогою конструктора __init__.
- у найпростішому випадку, клас може бути порожнім, просто декларувати щось нове, що буде використовуватись в майбутньому.

```python
class Student:
    pass  # порожній клас, який можна використовувати для створення об'єктів студентів у майбутньому

s1 = Student()  # створення об'єкта класу Student
print(type(s1))  # <class '__main__.Student'>
s1.name = "Богдан"  # динамічне додавання атрибуту name до об'єкта s1
s1.oop_mark = 5  # динамічне додавання атрибуту oop_mark до об'єкта s1
s1.math_mark = 4  # динамічне додавання атрибуту math_mark до об'єкта s1
s1.database_mark = 5  # динамічне додавання атрибуту database_mark до об'єкта s1

s2 = Student()  # створення ще одного об'єкта класу Student
s2.name = "Андрій"  # динамічне додавання атрибуту name до об'єкта s2
s2.oop_mark = 4  # динамічне додавання атрибуту oop_mark до об'єкта s2
s2.math_mark = 5  # динамічне додавання атрибуту math_mark до об'єкта s2
s2.micricontroller_mark = 4  # динамічне додавання атрибуту micricontroller_mark до об'єкта s2

# s1 s2 - належать до одного класу, але вони трохи різні, бо мають різні атрибути та значення цих атрибутів.

for student in [s1, s2]:  # ітерація по списку об'єктів студентів
    print(f"{student.name} отримав оцінки:")
    print(f"\tВища математика: {student.math_mark}")
    print(f"\tООП: {student.oop_mark}")
    print(f"\tБази даних: {getattr(student, 'database_mark', 'немає оцінки')}")
```

оскільки клас - це шаблон для створення обєктів - то завжди потрібно передбачати наперед, які атрибути та методи будуть потрібні для обєктів цього класу.
- в шаблоні класу - наперед передбачені атрибути створюються зад допомогою конструктора __init__ та методи, які будуть виконувати певні дії над обєктами цього класу.

```python
class Student:
    # Функції які знаходяться всередині класу називаються методами, 
    # а перший параметр завжди self - це посилання на об'єкт, який викликає цей метод.
    # дані які передаються в констурктор називаються аргументами 
    # і далі вони присвоюються атрибутам об'єкта, які створюються в конструкторі.
    def __init__(self, name, oop_mark, math_mark=0): # аргументи для створення об'єкта класу Student
        self.name = name # атрибут name об'єкта Student, який зберігає ім'я студента
        self.oop_mark = oop_mark # атрибут oop_mark об'єкта Student, який зберігає оцінку з ООП
        self.math_mark = math_mark # атрибут math_mark об'єкта Student, який зберігає оцінку з математики, за замовчуванням 0


sc1 = Student("Богдан", 5, 4) # створюємо конкретний обєкт пеердавши аргументи
sc2 = Student("Андрій", 4, 5) # створюємо конкретний обєкт пеердавши аргументи
# тепер ці обєкти є уніфікованими, бо вони створені за допомогою одного класу Student 
# і мають однакові атрибути name, oop_mark, math_mark.

for student in [sc1, sc2]:  # ітерація по списку об'єктів студентів
    print(f"{student.name} отримав оцінки:")
    print(f"\tВища математика: {student.math_mark}")
    print(f"\tООП: {student.oop_mark}")
```

- будь-який з атрибутів можна змінювати. 
- деколи це добре що ми можемо змінювати атрибути обєктів, але інколи це може призвести до небажаних наслідків. Тому важливо передбачати, які атрибути повинні бути доступні для зміни, а які ні.
- для читання існуґ інше поле - яке низивається властивості (property). Властивості дозволяють створювати атрибути, які можна читати, але не можна змінювати. Це дозволяє захистити дані обєкта від небажаних змін.

```python
class Student:
    def __init__(self, name):
        self.name = name # атрибут можна перевизначати

    @property
    def student_name(self):
        """Властивість доступна лише для читання, не можна змінювати""" 
        return f"Студент: {self.name}"  # повертає рядок з ім'ям студента

st1 = Student("Богдан")
st1.name = "Андрій"  # змінюємо атрибут name об'єкта st1
print(st1.student_name)  # викликаємо властивість student_name, яка повер
#st1.student_name = None
```

- також атрибути бувають приватними, публічними та захищеними. Приватні атрибути не можна змінювати ззовні класу, публічні атрибути можна змінювати ззовні класу, а захищені атрибути можна змінювати тільки в межах класу та його підкласів.
- приватні або захищені атрибути створюються за допомогою одинарного або подвійного підкреслення __ перед назвою атрибута, а публічні атрибути створюються без підкреслення.

```python
class Student:
    def __init__(self, name):
        self.name = name # публічний атрибут name, який можна змінювати
        self._protected_name = name # захищений атрибут _protected_name, 
        #який можна змінювати лише в межах класу та його підкласів
        #є правило що їх змінювати можна, але не бажано, бо вони призначені для внутрішнього використання класу
        self.__private_name = name # приватний атрибут __private_name, який не можна змінювати ззовні класу

    @property
    def get_private_name(self):
        """Властивість доступна лише для читання, не можна змінювати""" 
        return f"Приватний атрибут: {self.__private_name}"  # повертає рядок з ім'ям студента

st2 = Student("Богдан")
print(f"Початковий атрибут з конструктора:{st2.name}")  # виводимо значення атрибуту name об'єкта st2
print(f"Захищений атрибут: {st2._protected_name}")  # виводимо значення захищеного атрибуту _protected_name об'єкта st2

st2._protected_name = "Андрій"  # змінюємо значення захищеного атрибуту _protected_name об'єкта st2
print(f"Змінене поле: {st2._protected_name} - але юажано його не змінювати")  # виводимо нове значення захищеного атрибуту _protected_name об'єкта st2

#print(f"Тут буде помилка: {st2.__private_name}")
print(f"Властивість: {st2.get_private_name}")

st2.__private_name = "Анастісія"  # Це створило новий динамічний атрибут
print(f"Приватний атрибут: {st2.__private_name}")
print(f"Властивість: {st2.get_private_name}")  # виводимо значення властивості get_private_name об'єкта st2
```

- в класі ми можемо мати не моше статичні поля, але і виконувати певні обчислення або дії над обєктами цього класу. Для цього ми створюємо методи класу, які є функціями, що виконуються в межах класу та можуть взаємодіяти з атрибутами обєктів цього класу.

```python
class Student:
    def __init__(self, name: str|None = None, mark: int|None = None):
        self.name = name if name else "Без імені" # атрибут визначається через вираз
        #self.mark = # дія з обчисленнями робити сладно, тому використовуємо метод
        self.calculate_mark(mark) # викликаємо метод calculate_mark для обчислення оцінки

    def calculate_mark(self, init_mark: int|None = None) -> int:
        # дія з обчисленнями робити сладно, тому використовуємо метод
        # можливо тут складний алгоритм підрахунку оцінки
        if init_mark is None:
            self.mark = "Немає оцінки"
        elif 0 < init_mark < 50:
            self.mark = "Задовільно"
        elif 50 <= init_mark < 75:
            self.mark = "Добре"
        elif 75 <= init_mark <= 100:
            self.mark = "Відмінно"
        else:
            self.mark = "Некоректна оцінка"

so1 = Student("Богдан")
so2 = Student(mark=85)
so3 = Student("Анастісія", 45)
print(f"Студент 1: {so1.name}, оцінка: {so1.mark}")
print(f"Студент 2: {so2.name}, оцінка: {so2.mark}")
print(f"Студент 3: {so3.name}, оцінка: {so3.mark}")
```

- для прикладу напишемо малий клас який визначає персонаж в грі, який може атакувати та зціюватись (є методи які впливають на атрибути здоровя і змінюються його).
