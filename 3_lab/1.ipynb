## Обєктно-орієнтоване програмування (ООП)

- приклад-порівняння чому нам не достатньо. вбудованих класів та функцій, і чому нам потрібні власні класи та функції.
- описуємо студента та його оцінки.

```python
# Використовуємо вбудовані типи, словник, який є обєктом, що зберігає пари ключ-значення. 
student = {
    "Богдан": {"Вища математика": 4, "фізика": 5, "ООП": 5},
    "Андрій": {"Вища математика": 5, "фізика": 4},
    "Анастісія": {"Вища математика": 4, "фізика": 4},
}

print(type(student))  # <class 'dict'>

for name, grade in student.items():
    print(type(grade))  # <class 'dict'>
    print(f"{name} отримав оцінки:")
    print(f"\tВища математика: {grade['Вища математика']}")
    print(f"\tФізика: {grade['фізика']}")
    print(f"\tООП: {grade.get('ООП', 'немає оцінки')}\n") # метод get() дає значення за замовчуванням
```

- використання вбудованих типів для скласдих задач швидка стає громіздкою та незручною (не маштабованим).
- тому - ООП підхід дає можливість створювати власні класи та функції, які будуть більш зручними та маштабованими для вирішення складних задач.

- клас свторюється за допомогою ключового слова class, а обєкти створюються за допомогою конструктора __init__.
- у найпростішому випадку, клас може бути порожнім, просто декларувати щось нове, що буде використовуватись в майбутньому.

```python
class Student:
    pass  # порожній клас, який можна використовувати для створення об'єктів студентів у майбутньому

s1 = Student()  # створення об'єкта класу Student
print(type(s1))  # <class '__main__.Student'>
s1.name = "Богдан"  # динамічне додавання атрибуту name до об'єкта s1
s1.oop_mark = 5  # динамічне додавання атрибуту oop_mark до об'єкта s1
s1.math_mark = 4  # динамічне додавання атрибуту math_mark до об'єкта s1
s1.database_mark = 5  # динамічне додавання атрибуту database_mark до об'єкта s1

s2 = Student()  # створення ще одного об'єкта класу Student
s2.name = "Андрій"  # динамічне додавання атрибуту name до об'єкта s2
s2.oop_mark = 4  # динамічне додавання атрибуту oop_mark до об'єкта s2
s2.math_mark = 5  # динамічне додавання атрибуту math_mark до об'єкта s2
s2.micricontroller_mark = 4  # динамічне додавання атрибуту micricontroller_mark до об'єкта s2

# s1 s2 - належать до одного класу, але вони трохи різні, бо мають різні атрибути та значення цих атрибутів.

for student in [s1, s2]:  # ітерація по списку об'єктів студентів
    print(f"{student.name} отримав оцінки:")
    print(f"\tВища математика: {student.math_mark}")
    print(f"\tООП: {student.oop_mark}")
    print(f"\tБази даних: {getattr(student, 'database_mark', 'немає оцінки')}")
```

оскільки клас - це шаблон для створення обєктів - то завжди потрібно пеердбачати наперед, які атрибути та методи будуть потрібні для обєктів цього класу.
- в шаблоні класу - наперед передбачені атрибути створюються зад опомогою конструктора __init__ та методи, які будуть виконувати певні дії над обєктами цього класу.

```python
class Student:
    # Функції які знаходяться всередині класу називаються методами, 
    # а перший параметр завжди self - це посилання на об'єкт, який викликає цей метод.
    # дані які передаються в констурктор називаються аргументами 
    # і далі вони присвоюються атрибутам об'єкта, які створюються в конструкторі.
    def __init__(self, name, oop_mark, math_mark=0): # аргументи для створення об'єкта класу Student
        self.name = name # атрибут name об'єкта Student, який зберігає ім'я студента
        self.oop_mark = oop_mark # атрибут oop_mark об'єкта Student, який зберігає оцінку з ООП
        self.math_mark = math_mark # атрибут math_mark об'єкта Student, який зберігає оцінку з математики, за замовчуванням 0


sc1 = Student("Богдан", 5, 4) # створюємо конкретний обєкт пеердавши аргументи
sc2 = Student("Андрій", 4, 5) # створюємо конкретний обєкт пеердавши аргументи
# тепер ці обєкти є уніфікованими, бо вони створені за допомогою одного класу Student 
# і мають однакові атрибути name, oop_mark, math_mark.

for student in [sc1, sc2]:  # ітерація по списку об'єктів студентів
    print(f"{student.name} отримав оцінки:")
    print(f"\tВища математика: {student.math_mark}")
    print(f"\tООП: {student.oop_mark}")
```
